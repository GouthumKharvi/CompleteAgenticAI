Synchronus Programming

```python
import time 
```

```python
def fetch_weather():
    print("Fetching weather data...") #to feth weather data we have to wait for 4 seconds
    time.sleep(4)
    print("Weather data fetched successfully!")

def fetch_news():
    print("Fetching news data...") #to feth news data we have to wait for 2  seconds
    time.sleep(2)
    print("News data fetched successfully!")
    
#This 2 function is independent of each other.

def main():
    start_time = time.time()
    fetch_weather()                      #Although 2 function is independent we have to wait for previous function to complete before starting the next one. This is called synchronous programming.
    fetch_news()
    end_time = time.time()
    print(f"Total time taken: {end_time - start_time} seconds")


main()
```

Asyncgronus Programming

```python
import asyncio
import time
```

```python
async def fetch_weather():
    print("Fetching weather data...") #to feth weather data we have to wait for 4 seconds
    await asyncio.sleep(4)               #await is used to wait for the completion of the function without blocking the main thread.
    print("Weather data fetched successfully!")

async def fetch_news():
    print("Fetching news data...") #to feth news data we have to wait for 2  seconds
    await asyncio.sleep(2)
    print("News data fetched successfully!")

async def main():
    start_time = time.time()
    await asyncio.gather(fetch_weather(), fetch_news())
    end_time = time.time()
    print(f"Total time taken: {end_time - start_time} seconds")


await main()

#we are not waiting for the completion of the first function before starting the second one. Both functions are running concurrently, and we are waiting for both of them to complete before calculating the total time taken. This is called asynchronous programming.
```
